```python
import pandas as pd

url = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"

df = pd.read_csv(url)
```

Step 4 — Build your data dictionary

```python
import pandas as pd

data_dictionary = pd.DataFrame({
    "Column name": df.columns,
    "Type": [
        "Numeric" if pd.api.types.is_numeric_dtype(df[col])
        else "Categorical"
        for col in df.columns
    ],
    "Example value": [
        df[col].dropna().iloc[0] if df[col].notna().any() else None
        for col in df.columns
    ]
})

display(data_dictionary)
```

```python
meanings = {
    "customerID": "Unique customer identifier",
    "gender": "Customer's recorded gender",
    "SeniorCitizen": "Whether the customer is a senior citizen",
    "Partner": "Whether the customer has a partner",
    "Dependents": "Whether the customer has dependents",
    "tenure": "Number of months with the company",
    "PhoneService": "Whether the customer has phone service",
    "MultipleLines": "Whether the customer has multiple phone lines",
    "InternetService": "Type of internet service",
    "OnlineSecurity": "Whether online security is included",
    "OnlineBackup": "Whether online backup is included",
    "DeviceProtection": "Whether device protection is included",
    "TechSupport": "Whether technical support is included",
    "StreamingTV": "Whether streaming TV is included",
    "StreamingMovies": "Whether streaming movies are included",
    "Contract": "Customer's contract type",
    "PaperlessBilling": "Whether paperless billing is enabled",
    "PaymentMethod": "Customer's payment method",
    "MonthlyCharges": "Current monthly charges",
    "TotalCharges": "Total charges recorded for the customer",
    "Churn": "Whether the customer left the company"
}

data_dictionary.insert(
    1,
    "Real-life meaning",
    data_dictionary["Column name"].map(meanings)
)

display(data_dictionary)
```

```python
categorical_columns = [
    "customerID", "gender", "SeniorCitizen", "Partner",
    "Dependents", "PhoneService", "MultipleLines",
    "InternetService", "OnlineSecurity", "OnlineBackup",
    "DeviceProtection", "TechSupport", "StreamingTV",
    "StreamingMovies", "Contract", "PaperlessBilling",
    "PaymentMethod", "Churn"
]

data_dictionary.loc[
    data_dictionary["Column name"].isin(categorical_columns),
    "Type"
] = "Categorical"

display(data_dictionary)
```

### **Day 2. Explore the Data (EDA)**

1.1 Check the dataset shape

```python
print("Rows and columns:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

1.2 Check missing values in every column

```python
missing = pd.DataFrame({
    "Missing count": df.isnull().sum(),
    "Missing percentage": (df.isnull().mean() * 100).round(2)
})

display(missing)
```

1.3 Check empty text values

```python
empty_text = {}

for col in df.select_dtypes(include=["object"]).columns:
    empty_text[col] = df[col].apply(
        lambda x: isinstance(x, str) and x.strip() == ""
    ).sum()

display(pd.Series(empty_text, name="Empty text count"))
```

1.4 Check duplicate rows

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
display(df[df.duplicated(keep=False)].head(10))
```

1.5 Check strange values

```python
for col in df.select_dtypes(include=["object"]).columns:
    print(f"\n{col}")
    print(df[col].value_counts(dropna=False))
```

## Day 2 — Step 2, target balance

We need to understand how many customers churned before analyzing relationships between churn and other features.

```python
print("Churn counts:")
print(df["Churn"].value_counts())

print("\nChurn percentages:")
print((df["Churn"].value_counts(normalize=True) * 100).round(2))
```

The dataset contains 7,043 customers, of whom 26.54% churned and 73.46% stayed. The target classes are imbalanced, so accuracy alone may not be sufficient to evaluate the model.
